# Central Limit Theorem with Dice

One die is flat: every face is equally likely. But the **average of many dice** forms a bell curve. This notebook shows it by simulation, rolling dice 10,000 times for 1, 2, 5 and 30 dice.

**Key findings**
- The center of the bell stays at 3.5.
- More dice makes the bell narrower. The spread shrinks by about 1 / √n.

```python
%matplotlib inline
import numpy as np
import matplotlib.pyplot as plt

dice_counts = [1, 2, 5, 30]

fig, axes = plt.subplots(2, 2, figsize=(10, 8))

for ax, n_dice in zip(axes.flatten(), dice_counts):
    rolls = np.random.randint(1, 7, size=(10000, n_dice))
    averages = rolls.mean(axis=1)
    ax.hist(averages, bins=100)
    ax.set_title(f"Average of {n_dice} dice")

plt.suptitle("Central Limit Theorem: Dice", fontsize=14)
plt.tight_layout()
plt.savefig("clt_dice.png", dpi=150)
plt.show()
```

```python
import numpy as np

rolls = np.random.randint(1, 7, size=(10000, 30))
averages = rolls.mean(axis=1)

print("Mean of averages:", averages.mean())
print("Spread (std) of averages:", averages.std())
```

```python
for n_dice in [1, 5, 30]:
    rolls = np.random.randint(1, 7, size=(1000, n_dice))
    averages = rolls.mean(axis=1)
    print(n_dice, "dice --> spread", averages.std())
```
