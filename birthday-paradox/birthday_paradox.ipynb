# Birthday Paradox Simulator

"""in a room of 23 people, the chance that two share a birthday is about 50%.
This notebook checks that using both a formula (theory) and a simulation."""

```python
def theoretical_probability(n):
    "No one shares a birthday"
    
    prob_no_match = 1
    for i in range(n):
        prob_no_match *= (365 - i) / 365

    return 1 - prob_no_match
```

```python
print(theoretical_probability(23))
```

```python
import random 
def simulated_probability(n, trials=1000):
    matches = 0
    for _ in range(trials):
       birthdays = [random.randint(1, 365) for _ in range(n)]
       if len(birthdays) != len(set(birthdays)):
           matches += 1
       
    return matches / trials
```

```python
people = list(range(1, 61))

theory = [theoretical_probability(n) for n in people]
simulation = [simulated_probability(n) for n in people]
              
```

```python
plt.figure(figsize=(9, 5))
plt.plot(people, simulation, label= "Theory (formula)", linewidth=2)
plt.scatter(people, simulation, label="Simulation", color="orange", s=20)
plt.axhline(0.5, color="gray", linestyle="--", label="50% line")
plt.xlabel("Number of people in the room")
plt.ylabel("Probability of a shared birthday")
plt.title("Birthday Paradox")
plt.legend()
plt.show()
```
